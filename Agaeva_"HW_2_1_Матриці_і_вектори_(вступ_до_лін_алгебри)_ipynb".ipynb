```python
import numpy as np
```

**Рішення 1.1:**

```python
import random
```

```python

# створюємо пустий список
random_list_uniform = []
# Генеруєм 10 випадкових чисел от 0 до 1
for _ in range(10):
    random_list_uniform.append(random.uniform(0, 1))

# Перетворимо список в NumPy вектор
vector_uniform = np.array(random_list_uniform)

print("--- Варіант 1: random.uniform ---")
print("Вектор:\n", vector_uniform)
print("Розмірність вектора:", vector_uniform.shape)
print("Тип даних елементів:", vector_uniform.dtype)
```

```python
vector_random = np.random.rand(10)
print("--- Варіант 2:  np.random.rand ---")
print("Вектор:\n", vector_random)
print("Розмірність вектора:", vector_random.shape)
print("Тип даних елементів:", vector_random.dtype)
```

1. Задані дві матриці `A` і `B`розмірностями 3x2 та 2х3. Виконайте їхнє множення, щоб отримати матрицю 2x2. Виведіть результуючу матрицю `result_matrix` на екран.

2. Заданий вектор `C` розмірністю 2х1. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на вектор `C` і який результат буде. А потім виконайте множення і виведіть результат на екран.

3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

**Рішення 3.1:**

```python
result_matrix = np.dot(B, A)
result_matrix
```

**Рішення 3.2:**

Коли домножите result_matrix на вектор C то буде 1 х 2

```python
# має виконуватись правило - кількість колонок в першій матриці має дорівнювати кількості рядків у другій. Тобто можемо тільки result_matrix*C, навпаки не можемо
result_matrix1 = np.dot(result_matrix, C)
result_matrix1
```

**Рішення 3.3:**

Коли домножити result_matrix на матрицю D, то буде матриця розмірністю 2Х2

```python
result_matrix2 = np.dot(result_matrix, D)
result_matrix2
```
