1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
```

**Рішення 1.1.**

```python
# Визначаємо символічну змінну
x = sp.symbols('x')
```

```python
# Лінійна функція
f_lin = 10 * x - 5
derivative_f_lin = sp.diff(f_lin, x)
print(derivative_f_lin)
```

**Рішення 1.2.**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Подготовка данных для графика и построение

x_hours = np.linspace(0, 10, 100)

f_values = sp.lambdify(x, f_lin, 'numpy')(x_hours)

derivative_scalar_value = sp.lambdify(x, derivative_f_lin, 'numpy')(0) # Берем одно значение для проверки
derivative_values = np.full_like(x_hours, derivative_scalar_value) # Заполняем массив этой константой

# Вариант 2 (проще, если точно знаем, что производная константа):
# derivative_constant = float(derivative_f_lin) # Преобразуем символьную константу в число
# derivative_values = np.full_like(x_hours, derivative_constant)


# Создание графика
plt.figure(figsize=(10, 6))

plt.plot(x_hours, f_values, label='Оценка (f(x) = 10x - 5)', color='blue', linewidth=2)
plt.plot(x_hours, derivative_values, label="Производная f'(x) = 10", color='red', linestyle='--', linewidth=2)

plt.xlabel('Количество часов обучения (x)')
plt.ylabel('Оценка / Значение производной')
plt.title('Зависимость оценки от часов обучения и ее производная')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.show()
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
import math
```

**Рішення 2.1:**

```python
# Визначаємо символічну змінну
x2 = sp.symbols('x2')
#  функція
f2 = sp.sin(x2)
df2 = sp.diff(f2, x2) #похідна
```

```python
x_numeric = np.linspace(-2 * np.pi, 2 * np.pi, 400)
f2_numeric_values = sp.lambdify(x2, f2, 'numpy')(x_numeric)
df2_numeric_values = sp.lambdify(x2, df2, 'numpy')(x_numeric)
```

```python
# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x_numeric, f2_numeric_values, label='f(x2) = sin(x2)', color='blue')
plt.plot(x_numeric, df2_numeric_values, label="f'(x2) = cos(x2)", color='green')
plt.legend()
plt.show()
```

**Рішення 2.2.:**

```python
# Визначаємо символічну змінну
x22 = sp.symbols('x22')
#  функція
f22 = sp.log(x22)
df22 = sp.diff(f22, x22) #похідна
```

```python
x_numeric2 = np.linspace(-2 * np.pi, 2 * np.pi, 400)
f22_numeric_values = sp.lambdify(x22, f22, 'numpy')(x_numeric2)
df22_numeric_values = sp.lambdify(x22, df22, 'numpy')(x_numeric2)
```

```python
# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x_numeric2, f22_numeric_values, label='f(x22) = ln(x22)', color='blue')
plt.plot(x_numeric2, df22_numeric_values, label="f'(x22) = 1/x22", color='green')
plt.legend()
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
# Визначаємо символічну змінну
x3 = sp.symbols('x3')
#  функція
f = x3**2 - 5*x3+1
df = sp.diff(f, x3) #похідна
x_quantity = np.linspace(0, 100, 100)

f_q = sp.lambdify(x3, f, 'numpy')(x_quantity)

derivative_scalar_value = sp.lambdify(x3, df, 'numpy')(0) # Берем одно значение для проверки
derivative_values = np.full_like(x_quantity, derivative_scalar_value) # Заполняем массив этой константой

# Создание графика
plt.figure(figsize=(10, 6))

plt.plot(x_quantity, f_values, label='Оцінка (f(x3) = x3**2 - 5*x3+1)', color='blue', linewidth=2)
plt.plot(x_quantity, derivative_values, label="Похідна f'(x3) = 2*x3-5", color='red', linestyle='--', linewidth=2)

plt.xlabel('кількість вироблених одиниць (x)')
plt.ylabel('Оцінка / Значення похідної')
plt.title('Залежність кіьлкості виробництва та її похідної')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.show()
```
